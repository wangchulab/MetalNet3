```python
# input
diverged_cluster = "./data/repId-cliques-numAnnoPro-isAnno-memAnnoLevel-repLen-repName-avgPLDDT.tsv"
# output
result_file = "./data/selection.tsv"
```

```python
import pandas as pd

df = pd.read_table(diverged_cluster, header=None, names=["rep_id", "cliques", "num_anno_pro", "is_anno", "mem_anno_level", "rep_len", "rep_name", "avg_plddt"])
len(df)

df = df[df['mem_anno_level'] != 1.0]
len(df)

df = df[df['rep_len'].map(lambda x: x >= 100 and x <= 600)]
len(df)

df = df[df['avg_plddt'].map(lambda x: x >= 70)]
len(df)

# df = df[df['num_anno_pro'] == 0]
# len(df)

df = df[df['cliques'].map(lambda x: all([len(i.split(",")) >= 2 for i in x.split(";")]))]
len(df)
```

```python
df.to_csv(result_file, sep="\t", index=None)
```
