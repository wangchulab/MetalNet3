```python
# input
data = {
    'AFDB6': [0.332, 0.143, 0.896, 0.170], # order: ni, cu, zn, fe
    'AFDB7': [0.085, 0.207, 1.618, 0.100]
}
metals = ['Ni', 'Cu', 'Zn', 'Fe']
# output
fig_file = "./fig/afdb6-7_icp-mass.pdf"
```

```python
import matplotlib.pyplot as plt


plt.style.use(f"{PROJECT_DIR}/asset/metalnet.mplstyle")

fig, ax = plt.subplots(figsize=(2.4, 2.4))
bar_width = 0.3
group_spacing = 3 * bar_width
positions = []

for group_idx, (group_name, values) in enumerate(data.items()):
    x_start = group_idx * (len(metals)*bar_width + group_spacing)
    for metal_idx, (metal, value) in enumerate(zip(metals, values)):
        x = x_start + metal_idx * bar_width
        ax.bar(x, value, width=bar_width,
               label=metal if group_idx == 0 else "",
               edgecolor='black')
    positions.append(x_start + (len(metals)-1)*bar_width/2)

ax.set_xticks(positions)
ax.set_xticklabels(data.keys())
ax.set_ylabel('Ratio in mol (ion / protein)')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.legend(metals, bbox_to_anchor=(0.5, 1))
plt.savefig(fig_file, transparent=True, bbox_inches='tight')
```
