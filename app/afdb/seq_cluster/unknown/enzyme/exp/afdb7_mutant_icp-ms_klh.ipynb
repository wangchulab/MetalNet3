```python
# input
data = { # Zn ion
    "WT": [1.646155, 1.704949, 1.654041], 
    "2A-1": [1.330637, 1.222881, 1.135564], # C237A & C240A
    "2A-2": [0.862717, 0.674354, 0.604507], # C272A & C275A
    "4A":  [0.544332, 0.52232, 0.302085],
    "8A": [0.037004, 0.035303, 0.036266],
    "NT": [0.066534, 0.067527, 0.069216]
}
# output
fig_file = "./fig/afdb7_mutant_icp-mass_klh.pdf"
```

```python
import numpy as np

categories = data.keys()
means = [np.mean(data[cat]) for cat in categories]
means
stds = [np.std(data[cat], ddof=1) for cat in categories]
```

```python
import matplotlib.pyplot as plt


plt.style.use(f"{PROJECT_DIR}/asset/metalnet.mplstyle")
color_cycle = plt.rcParams['axes.prop_cycle'].by_key()['color']

fig, ax = plt.subplots(figsize=(2.4, 2.4))

x = np.arange(len(categories))
bar_width = 0.5
bars = ax.bar(x, means, width=bar_width, color=color_cycle, yerr=stds)

for i, cat in enumerate(categories):
    y_vals = data[cat]
    ax.scatter(np.full(len(y_vals), x[i]), y_vals, s=1, c="black")

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.set_ylabel("Ratio in mol (ion / protein)")
plt.yticks([i / 10 for i in range(0, 21, 3)])
plt.savefig(fig_file, transparent=True, bbox_inches='tight')
```
