```python
# input
pred_file = "./tmp/no_anno_pred.tsv"
fasta_file = "./tmp/no_anno.fasta"

# output
filtered_file = "./data/filtered.tsv"
```

```python
import pandas as pd
df = pd.read_table(pred_file)
```

```python
from Bio import SeqIO
id2seq = dict()
for r in SeqIO.parse(fasta_file, "fasta"):
    id2seq[r.id.split("-")[1]] = str(r.seq)

df['len'] = df['seq_id'].map(lambda x: len(id2seq[x]))
```

### length: 100 - 600
### metal group type: transition metal (1)
### proba: all > 0.5

```python
df = df[df['len'].map(lambda x: x >= 100 and x <= 600)]
len(df)

df = df[df['metal_group_type'].map(lambda x: x.count("1") >= 3)]
len(df)

df = df[df['proba'].map(lambda x: all(float(r) > 0.5 for r in x.split(",")))]
len(df)
```

```python
df.to_csv(filtered_file, sep="\t", index=None)
```
