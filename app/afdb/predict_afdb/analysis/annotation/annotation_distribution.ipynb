```python
# output
fig_file="./fig/metaldb_annotation_source.pdf"
```

```python
x = ["AFDB50", "AFDB clusters", "TED", "Pfam", "combined"]
total = 38361041
y = [16311554, 10096783, 5931519, 8709939, 27927308]
y = [i / total for i in y]
y_no_anno = [1 - i for i in y]
y
```

```python
import matplotlib.pyplot as plt

plt.style.use(f"{PROJECT_DIR}/asset/metalnet.mplstyle")
plt.figure(figsize=(1.8, 2.4))

plt.bar(x, y, width=0.6)
plt.bar(x, y_no_anno, width=0.6, bottom=y)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.xticks(rotation=45, ha="right")
ax.set_ylabel("Percentage of proteins in MetalDB")
plt.legend(
    labels=['w/ annotation', 'w/o annotation'],
    loc='lower center',
    bbox_to_anchor=(0.5,1)
)
plt.savefig(fig_file, bbox_inches="tight", transparent=True)
```
