```python
y_arc = [44727, 117039, 29219]
y_bac = [57488, 113832, 18052]
y_euk = [27338, 166137, 2433]

y_arc = [i / 200000 for i in y_arc]
y_bac = [i / 200000 for i in y_bac]
y_euk = [i / 200000 for i in y_euk]

y_arc
y_bac
y_euk

x = ["Alkaline metals", "Transition metals", "Fe-S clusters"]

sum(y_arc)
sum(y_bac)
sum(y_euk)
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use(f"{PROJECT_DIR}/asset/metalnet.mplstyle")

plt.figure(figsize=(2.4, 2.4))

plt.ylabel("Proportion")
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

coor_x = np.arange(len(x))
plt.xticks(coor_x, labels=x, rotation=90)

width = 0.2
plt.bar(coor_x - width, y_arc, width=width)
plt.bar(coor_x, y_bac, width=width)
plt.bar(coor_x + width, y_euk, width=width)

plt.legend(
    ['Archaea', 'Bacteria', 'Eukaryota'], 
    bbox_to_anchor=(0.5,1), 
    loc='upper right', 
)
plt.savefig(fig_file, bbox_inches="tight", transparent=True)
```
