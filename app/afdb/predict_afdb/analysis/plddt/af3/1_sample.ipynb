```python
# input
pred_file = "../../../data/pred_site_plddt_proba.tsv"
# output
sampled_file = "./data/sampled.tsv"
```

```python
import pandas as pd
df = pd.read_table(pred_file)
```

```python
import numpy as np
uppers = [50, 70, 90, 100]
lowers = [0, 50, 70, 90]
def assign_bin(plddts_str):
    plddt = np.median([float(i) for i in plddts_str.split(",")])
    for i, j in zip(lowers, uppers):
        if plddt >= i and plddt < j:
            return f"{i}-{j}"

df["bin"] = df['plddt'].map(lambda x: assign_bin(x))
```

```python
sampled_dfs = []
for (bin,), df_bin in df.groupby(by=['bin']):
    sampled_dfs.append(df_bin.sample(n=60, random_state=42))
pd.concat(sampled_dfs).to_csv(sampled_file, sep="\t", index=None, header=None, columns=["seq_id", "bin"])
```
