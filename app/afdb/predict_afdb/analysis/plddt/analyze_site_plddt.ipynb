```python
# input
site_plddt_file = "../../data/pred_site_plddt_proba.tsv"
# output
fig_file = "./fig/site_plddt_pie.pdf"
```

```python
import pandas as pd
import numpy as np

df = pd.read_table(site_plddt_file)
site_plddts = df['plddt'].map(lambda x: np.median([float(i) for i in x.split(",")])).tolist()
```

```python
cnts, edges = np.histogram(site_plddts, bins=[0, 50, 70, 90, 100])
cnts
edges
```

```python
ratio = cnts / cnts.sum()
ratio
```

```python
ratio = [0.02214349, 0.04061813, 0.24480482, 0.69243355]
```

```python
import matplotlib.pyplot as plt

plt.style.use(f"{PROJECT_DIR}/asset/metalnet.mplstyle")

orig_colors = ["#FF7D45", "#FFDB13", "#65CBF3", "#0053D6"]
pastel_colors = ["#FFB396", "#FFF099", "#A5DCF5", "#9FB9F2"]

plt.figure(figsize=(3, 3))
plt.pie(ratio, autopct='%1.1f%%', colors=pastel_colors)
plt.savefig(fig_file, bbox_inches="tight", transparent=True)
```
