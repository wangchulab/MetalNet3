```python
# input
no_anno_homomer = "../tmp/pred_no_anno.tsv"
selected = "./tmp/no_anno_struct_checked.tsv"
seq_file = "../tmp/no_anno_homomer.fasta"
# output
exp_candidates = "./tmp/exp_candidates.tsv"
```

```python
import pandas as pd


df = pd.read_table(no_anno_homomer)
struct_checked_ids = set(pd.read_table(selected, header=None)[0])
df = df[df['seq_id'].map(lambda x: x in struct_checked_ids)]
len(df)
```

```python
# no pf
df = df[df['org'].map(lambda x: x != "pf")]
len(df)
```

```python
from Bio import SeqIO

id2seq = dict()
for i in SeqIO.parse(seq_file, "fasta"):
    id2seq[i.id] = str(i.seq)
df['len'] = df['seq_id'].map(lambda x: len(id2seq[x]))
df = df[df['len'].map(lambda x: x <= 800)]
len(df)
```

```python
df.drop(columns=['pfam_id', 'ted_id', 'anno_level']).to_csv(exp_candidates, sep="\t", index=None)
```
