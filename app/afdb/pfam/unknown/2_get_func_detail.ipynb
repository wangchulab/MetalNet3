```python
# input
no_anno_pfam = "./data/no_anno_pfam.tsv"
# output
func_detail = "./tmp/func_pfam_detail.tsv"
```

```python
import pandas as pd

df = pd.read_table(no_anno_pfam)
len(df)
df = df[df['func_unknown'] == False]
df.drop(columns=['func_unknown'], inplace=True)
len(df)
```

```python
import requests

API_TEMPLATE = "https://www.ebi.ac.uk/interpro/api/entry/pfam/{pfam_id}/"
HEADERS = {
    "User-Agent": "get-pfam-description/1.0 (+https://your.email.or.repo)",
    "Accept": "application/json"
}

def get_pfam_detail(pfam_id: str, timeout: int = 10):
    url = API_TEMPLATE.format(pfam_id=pfam_id)
    r = requests.get(url, headers=HEADERS, timeout=timeout)
    data = r.json()['metadata']['description']
    detail = "\n".join([i['text'] for i in data])
    return detail
```

```python
import time
results = []
for _, row in df.iterrows():
    pfam_id = row['pfam_id']
    try:
        detail = get_pfam_detail(pfam_id)
        time.sleep(2)
    except:
        detail = ""
        print(f"failed for {pfam_id}")
    results.append({
        "pfam_id": pfam_id,
        "detail": detail 
    })
```
