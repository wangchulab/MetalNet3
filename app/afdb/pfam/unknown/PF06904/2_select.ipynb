```python
# input
pred_info = "./tmp/PF06904_pred.tsv"
seq_file = "./tmp/PF06904.fasta"
```

```python
import pandas as pd
from Bio import SeqIO

df_pred = pd.read_table(pred_info)

id2seq = dict()
for i in SeqIO.parse(seq_file, "fasta"):
    id2seq[i.id.split("-")[1]] = str(i.seq)

df_pred['resi'] = df_pred.apply(lambda row: ",".join([id2seq[row['seq_id']][int(i) - 1] for i in row['pred_seq_num'].split(",")]), axis=1)
```

```python
from typing import Counter


residues = [i for j in df_pred['resi'] for i in j.split(",")]
resi2cnt = Counter(residues)
total = len(residues)
resi2ratio = {k: v / total for k, v in resi2cnt.items()}
resi2cnt
resi2ratio
```
