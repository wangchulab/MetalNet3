```python
# input
go_obo_file = "../../_database/go/go.obo"
mbp_go_terms = ['GO:0046872', 'GO:0051540']
enzyme_go_terms = ["GO:0003824"]
membrane_go_terms = ["GO:0016020"]

# output
output_mbp_go_members = "./data/mbp_go_id-name.tsv"
output_enzyme_go_members = "./data/enzyme_go_id-name.tsv"
output_membrane_go_members = "./data/membrane_go_id-name.tsv"
```

```python
from goatools.obo_parser import GODag
import pandas as pd


def extract_go_members(
    go_dag: GODag,
    ancestor_go_terms: list[str],
):
    targets = set()
    for g in ancestor_go_terms:
        targets.add(g)
        all_children = go_dag[g].get_all_children()
        targets |= all_children

    records = []
    for g in targets:
        records.append({"go_id": g, "name": go_dag[g].name})

    return pd.DataFrame(records).sort_values("go_id")

go_dag = GODag(go_obo_file, prt=None)
```

```python
df = extract_go_members(go_dag, mbp_go_terms)
df.to_csv(output_mbp_go_members, sep="\t", index=None, header=None)

df = extract_go_members(go_dag, enzyme_go_terms)
df.to_csv(output_enzyme_go_members, sep="\t", index=None, header=None)

df = extract_go_members(go_dag, membrane_go_terms)
df.to_csv(output_membrane_go_members, sep="\t", index=None, header=None)
```
