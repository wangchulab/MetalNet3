```python
# input
cys_mp_file = "./data/4140_mbp_cysMP.tsv"
pred_file = "../afdb/predict_afdb/data/pred/pred_ge_3_clique_3.tsv"
human = "../afdb/_database/afdb_clusters/3-sapId_sapGO_repId_cluFlag_LCAtaxId.tsv"
```

```python
import pandas as pd

df_pred = pd.read_table(pred_file)
df_pred['seq_id'] = df_pred['seq_id'].map(lambda x: x.split("-")[1])

human_ids = set(pd.read_table(human, header=None)[0])
df_pred = df_pred[df_pred['seq_id'].map(lambda x: x in human_ids)]
del human_ids
len(df_pred)
```

```python
cys_map = pd.read_table(cys_mp_file)
len(set(cys_map['seq_id']))
pro_inter =  set(df_pred['seq_id']) & set(cys_map['seq_id'])
len(pro_inter)
```

```python
cys_map_resi = set(zip(cys_map['seq_id'], cys_map['seq_num'].map(lambda x: x - 1)))
pred_resi = set()
for _, row in df_pred.iterrows():
    seq_id = row['seq_id']
    posis = list(map(int, row['posi'].split(",")))
    for p in posis:
        pred_resi.add((seq_id, p))
len(cys_map_resi)
len(pred_resi)
len(pred_resi & cys_map_resi)
```
