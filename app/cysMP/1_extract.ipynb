```python
# input
cysmp_raw = "./tmp/cysMP.txt"
# output
cysmp = "./data/4140_mbp_cysMP.tsv"
```

```python
import pandas as pd
```

```python
metal_types = {"Zn", "Cu", "Ca", "Co", "Ni", "Mn", "FeIII", "FeII", "Sb", "Cr", "Sn"}

df = pd.read_table(cysmp_raw)
records = []
for _, row in df.iterrows():
    seq_id = row['Uniprot Accession']
    resi_seq_num = int(row['Iodo-APB modified site'][1:])
    ok_metals = []
    for m in metal_types:
        try:
            p = float(row[f'P_{m}'])
            log = float(row[f'Log2({m})'])
            if p < 0.05 and log >= 1:
                ok_metals.append(m[:2])
        except: 
            continue
    if len(ok_metals) > 0:
        records.append({
            "seq_id": seq_id,
            "seq_num": resi_seq_num,
            "metal_types": ",".join(ok_metals)
        })
```

```python
df = pd.DataFrame(records)
len(set(df['seq_id']))
df.to_csv(cysmp, sep="\t", index=None)
```
