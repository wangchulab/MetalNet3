```python
# input
metaldb = "../afdb/predict_afdb/data/pred/pred_ge_3_clique_3.tsv"
mbpa_dir = "./data/"
```

```python
import pandas as pd

metals = {"Mg", "Ca", "Mn", "Fe", "Co", "Ni", "Cu", "Zn"}
metal_ids = dict()
for m in metals:
    df = pd.read_csv(mbpa_dir + f"./{m}.csv")
    metal_ids[m] = set(df['accession'])

df_pred = pd.read_table(metaldb)
pred_ids = set(df_pred['seq_id'].map(lambda x: x.split("-")[1]))
```

```python
records = []
for i in metals:
    total = len(metal_ids[i])
    inter = len(metal_ids[i] & pred_ids)
    records.append({
        "total": total,
        "intersection": inter,
        "ratio": inter / total,
        "type": i
    })
```

```python
pd.DataFrame(records).sort_values(by="ratio")
```
